```python
df2['time'] = pd.to_datetime(df2['Timestamp IST'], format='%d-%m-%Y %H:%M')
df2['date'] = df2['time'].dt.date
```

```python
df2.drop(columns=['Timestamp IST','time'],inplace=True)
```

```python
df1['datetime'] = pd.to_datetime(df1['timestamp'], unit='s')
df1['date'] = df1['datetime'].dt.date
df1.drop(columns=['timestamp','datetime'],inplace=True)
```

```python
df1.head()
```

```python
df2.head()
```

```python
df=df2.merge(df1,on='date',how='left')
```

```python
df
```

```python
df.shape
```

```python
df.info()
```

```python
df = df.dropna(subset=['classification'])
```

```python
daily = df.groupby(['Account', 'date','classification'])['Closed PnL'].sum().reset_index()
daily
```

```python
daily['win'] = daily['Closed PnL'] > 0
daily
```

```python
daily.groupby('classification')['win'].mean()
```

```python
daily.groupby('Account')['win'].mean()
```

```python
order_size = df.groupby(['Account', 'date', 'Order ID'])['Size USD'].sum().reset_index()
avg_size = order_size.groupby(['Account', 'date'])['Size USD'].mean().reset_index(name='avg_trade_size')
daily = daily.merge(avg_size, on=['Account','date'])
```

```python
daily.head()
```

```python
daily['win_rate_per_trader'] = daily.groupby('Account')['win'].transform('mean')
daily.groupby('classification').agg({
    'Closed PnL': 'mean',
    'win': 'mean',
    'avg_trade_size': 'mean'
})
```
